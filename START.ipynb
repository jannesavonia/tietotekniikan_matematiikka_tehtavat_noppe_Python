# Kurssin käynnistys
Paina **START**. Ensimmäisellä käyttökerralla kurssin tiedostot kopioidaan omaan `~/my-work`-hakemistoosi. Myöhemmin START avaa olemassa olevan työskentelykopion.


```python
from pathlib import Path
import sys
sys.path.insert(0, str(Path.cwd()))
from start_course import start_course
import ipywidgets as widgets
from IPython.display import display, clear_output

button = widgets.Button(description="START", button_style="success", icon="play", layout=widgets.Layout(width="180px", height="55px"))
output = widgets.Output()

def on_start_clicked(_):
    button.disabled = True
    with output:
        clear_output()
        try:
            start_course()
        except FileExistsError as e:
            print("KURSSIA EI KÄYNNISTETTY")
            print(e)
            button.disabled = False
        except Exception as e:
            print("VIRHE")
            print(e)
            button.disabled = False

button.on_click(on_start_clicked)
display(button, output)
```
